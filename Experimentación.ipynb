```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

# Reiniciar los índices de los DataFrames y Series resultantes
X_train.reset_index(drop=True, inplace=True)
X_test.reset_index(drop=True, inplace=True)
y_train.reset_index(drop=True, inplace=True)
y_test.reset_index(drop=True, inplace=True)
```

```python
# Tratamiento especial para el caso en el que las categoricas tengan muchas categorias distintas.
# Quiero que en ese caso se agrupen en una sola categoría llamada RESTO si su frecuencia es menor al 1% en train, y aplico eso mismo en test
# Definimos cómo queremos llamar a los nulos

etiqueta_nulos = 'MISSING'
etiqueta_resto = 'RESTO'
umbral = 0.01
mapa_categorias_frecuentes = {}

for col in variables_categoricas:
    # PASO PREVIO: Rellenar nulos antes de calcular nada
    # Lo hacemos en Train y Test para evitar errores, aunque el cálculo de frecuencias sea solo en Train
    X_train[col] = X_train[col].fillna(etiqueta_nulos)
    X_test[col] = X_test[col].fillna(etiqueta_nulos)
    
    # IMPORTANTE: Asegurarnos de que son strings (a veces los nulos mezclan tipos)
    X_train[col] = X_train[col].astype(str)
    X_test[col] = X_test[col].astype(str)
    
    # 1. Calcular frecuencias SOLO en Train
    conteo = X_train[col].value_counts(normalize=True)
    
    # 2. Obtener etiquetas frecuentes
    etiquetas_frecuentes = conteo[conteo >= umbral].index.tolist()

    mapa_categorias_frecuentes[col] = etiquetas_frecuentes
    
    # 3. Aplicar agrupación
    X_train[col] = np.where(X_train[col].isin(etiquetas_frecuentes), X_train[col], etiqueta_resto)
    X_test[col] = np.where(X_test[col].isin(etiquetas_frecuentes), X_test[col], etiqueta_resto)
```

```python
# 1. Configurar el Imputador
imputer = SimpleImputer(strategy='mean')

# 2. Ajustar el imputador SOLO con los datos de entrenamiento
# Esto "aprende" cuál es la media de cada columna
imputer.fit(X_train_num)

# 3. Transformar ambos conjuntos
# Rellenamos los huecos con la media aprendida en el paso anterior
X_train_num_imputed = imputer.transform(X_train_num)
X_test_num_imputed = imputer.transform(X_test_num)

# Crear el modelo de regresión logística
modelo_reg_log = LogisticRegression(penalty=None)

# Ajustar el modelo con los datos de entrenamiento
modelo_reg_log.fit(X_train_num_imputed, y_train)

# Predecir en los datos de prueba
predicciones_reg_log = modelo_reg_log.predict(X_test_num_imputed)

# Evaluar el modelo
accuracy_reg_log = modelo_reg_log.score(X_test_num_imputed, y_test)
print("Exactitud del modelo:", accuracy_reg_log)

# Predecir probabilidades en lugar de clases
predicciones_prob_reg_log = modelo_reg_log.predict_proba(X_test_num_imputed)[:, 1]

# Calcular el AUC
auc_reg_log = roc_auc_score(y_test, predicciones_prob_reg_log)
print("*******************************************************************************************************************************")
print("PARA EL ARTÍCULO:")
print("Área bajo la curva ROC (AUC):", auc_reg_log)
print("*******************************************************************************************************************************")
```

### Regresión logística a todas las variables

```python
X_train_cat = X_train[variables_categoricas].fillna('MISSING')
X_test_cat = X_test[variables_categoricas].fillna('MISSING')

# Convertir las variables categóricas a tipo 'str'
for col in variables_categoricas:
    X_train_cat.loc[:, col] = X_train_cat[col].astype('str')
    X_test_cat.loc[:, col] = X_test_cat[col].astype('str')
```

```python
# --- PASO 1: Tratar variables NUMÉRICAS ---
# Imputamos la media
imputer_num = SimpleImputer(strategy='mean')
# fit_transform en train, solo transform en test
X_train_num_clean = imputer_num.fit_transform(X_train_num)
X_test_num_clean = imputer_num.transform(X_test_num)

# --- PASO 2: Tratar variables CATEGÓRICAS (One Hot Encoding) ---
# handle_unknown='ignore' es VITAL: si en el test aparece una categoría nueva
# que no estaba en el train, el modelo la ignorará en lugar de dar error.
# sparse_output=False nos devuelve un array normal (no una matriz dispersa)
ohe = OneHotEncoder(handle_unknown='ignore', sparse_output=False)

X_train_cat_enc = ohe.fit_transform(X_train_cat)
X_test_cat_enc = ohe.transform(X_test_cat)

# --- PASO 3: Unir (Concatenar) los datos ---
# Usamos numpy horizontal stack (hstack) para poner las columnas numéricas
# al lado de las categóricas
X_train_total = np.hstack((X_train_num_clean, X_train_cat_enc))
X_test_total = np.hstack((X_test_num_clean, X_test_cat_enc))

# --- PASO 4: Entrenar el modelo ---
modelo_reg_log = LogisticRegression(penalty=None)
modelo_reg_log.fit(X_train_total, y_train)
```

```python
# Predecir en los datos de prueba
predicciones_reg_log = modelo_reg_log.predict(X_test_total)

# Evaluar el modelo
accuracy_reg_log = modelo_reg_log.score(X_test_total, y_test)
print("Exactitud del modelo:", accuracy_reg_log)

# Predecir probabilidades en lugar de clases
predicciones_prob_reg_log = modelo_reg_log.predict_proba(X_test_total)[:, 1]

# Calcular el AUC
auc_reg_log = roc_auc_score(y_test, predicciones_prob_reg_log)
print("*******************************************************************************************************************************")
print("PARA EL ARTÍCULO:")
print("Área bajo la curva ROC (AUC):", auc_reg_log)
print("*******************************************************************************************************************************")
```

```python
coefficients_reg_log = modelo_reg_log.coef_

intercept_reg_log = modelo_reg_log.intercept_

print("Coeficientes:", coefficients_reg_log)
print("Intercept:", intercept_reg_log)
```
